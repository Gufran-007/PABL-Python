```python
import kagglehub
path = kagglehub.dataset_download("uciml/adult-census-income")

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv(f"{path}/adult.csv")
income_related = df["fnlwgt"]

#1.
print(income_related.mean())
print(income_related.median())
print(income_related.mode())

#2.
print(income_related.min())
print(income_related.max())
range_val = income_related.max() - income_related.min()
print(range_val)

#3.
income_var = income_related.var()
print(f"Variance:{income_var}")
income_std = income_related.std()
print(f"Variance:{income_std}")

#4.
q1, q2, q3 = income_related.quantile([0.25, 0.50, 0.75])
print(f"q1 = {q1}, q2 = {q2}, q3 = {q3}")

#5. Outlier
iqr = q3-q1
lower_limit = q1-1.5*iqr
upper_limit = q3+1.5*iqr
outliers = income_related[(income_related < lower_limit) | (income_related > upper_limit)]
print(len(outliers))


```
